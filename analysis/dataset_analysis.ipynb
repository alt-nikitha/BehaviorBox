```python
import json
import pandas as pd

with open("/data/user_data/nsrikant/bbox_data/data/olmo_validation_texts.jsonl", "r") as f:
    lines = f.readlines()

records = []
for line in lines:
    data = json.loads(line)
    records.append(data)

df = pd.DataFrame.from_records(records)
print(df.shape)
print(df.columns)
df["dataset"].value_counts().plot(kind="bar", title="Number of samples per domain")
```

```python
import json
import pandas as pd

with open("/data/user_data/nsrikant/bbox_data/data/amber_300_unseen.jsonl", "r") as f:
    lines = f.readlines()

records = []
for line in lines:
    data = json.loads(line)
    records.append(data)

df = pd.DataFrame.from_records(records)
print(df.shape)
print(df.columns)
# df["domain"].value_counts().plot(kind="bar", title="Number of samples per domain")
import matplotlib.pyplot as plt

# 1. Calculate counts
domain_counts = df["domain"].value_counts()

# 2. Set a very tall figure height to accommodate all rows (e.g., 20 inches)
# Adjust the '20' based on how many total domains you have
plt.figure(figsize=(12, 20)) 

# 3. Plot horizontal bars
domain_counts.plot(kind="barh", color="steelblue", edgecolor="black")

# 4. Invert y-axis so the largest domain is at the top
plt.gca().invert_yaxis()

# 5. Add labels and styling
plt.title("Number of Samples per Domain (Full Distribution)", fontsize=16, pad=20)
plt.xlabel("Sample Count", fontsize=12)
plt.ylabel("Domain", fontsize=12)
plt.grid(axis='x', linestyle='--', alpha=0.6)

# 6. Crucial: ensure labels have enough space
plt.tight_layout()

plt.show()
```
